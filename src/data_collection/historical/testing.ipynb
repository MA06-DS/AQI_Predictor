```python
import pandas as pd
aqi = pd.read_csv(r"..\..\..\data\historical\karachi_pm25_all.csv")

print(aqi.head())
len(aqi)
print(aqi["pm25"].isnull().sum())
```

```python
import pandas as pd
meteo = pd.read_csv(r"..\..\..\data\historical\karachi_weather_all.csv")
print(meteo.head())
len(meteo)
```

```python
print(aqi["datetime_local"].dtype)
print(meteo["datetime_local"].dtype)
```

```python
aqi["datetime_local"] = pd.to_datetime(
    aqi["datetime_local"],
    format="mixed"
)

meteo["datetime_local"] = pd.to_datetime(
    meteo["datetime_local"],
    format="mixed"
)
aqi["datetime_local"] = aqi["datetime_local"].dt.tz_localize(None)
```

```python
print(aqi["datetime_local"].dtype)
print(meteo["datetime_local"].dtype)
```

```python
print("Weather:")
print(meteo["datetime_local"].min(), meteo["datetime_local"].max())

print("\nPM25:")
print(aqi["datetime_local"].min(), aqi["datetime_local"].max())
```
